```python
from bam_masterdata.logger import logger
from bam_masterdata.metadata.entities import CollectionType
from bam_masterdata.cli.run_parser import RunParsers
from pathlib import Path
from pybis import Openbis
from decouple import config as environ
import getpass


coll = CollectionType()
```

Run File for a bam-masterdata OpenBis Parser

Step 1:
Import the Parser Class from the src folder.

```python
from src.example import ExampleParser

parser = ExampleParser()
```

Step 2: Declare the Files to parse

```python

FILE_1 = ""
#FILE_2 = ...

files = [FILE_1] # Change [FILE_1] to [FILE_1,FILE_2,...] for multiple files
```

Step 3:
Set OpenBis Url and input password.
- Set Openbis URL
- Input own username and password

```python
URL = "https://your-openbis-instance.com" # openBIS URL
print(f"Connecting to openBIS instance at {URL}...")
o = Openbis(URL)

username = input("Username: ")
password = getpass.getpass("Password: ")
o.login(
    username,
    password,
    save_token=True,
)
```

Step 4:
Define Space, Projekt and optionally Collection for OpenBis

```python
SPACE = "EXAMPLE_SPACE"
PROJECT = "EXAMPLE_PROJECT"
O_COLLECTION = "EXAMPLE_COLLECTION" # optional

#if not provided, the file are save in the project
```

Step 5:
Define Parser run.

```python
parse = {parser: files}

parsing = RunParsers(o,SPACE,PROJECT,O_COLLECTION,parse,logger=logger,)
```

Step 6:
Starting the run.

```python

print("Starting parsing...")
parsing.run()
print("done")
```
